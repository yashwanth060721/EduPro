```python
import sys
!{sys.executable} -m pip install pandas numpy plotly

import pandas as pd
import numpy as np
import plotly.express as px

# 1. Create Mock Data (matching your internship dataset structure)
np.random.seed(42)
n_users = 1000
n_courses = 20
n_transactions = 3500

# Users Table
users = pd.DataFrame({
    "UserID": [f"USR_{1000 + i}" for i in range(n_users)],
    "UserName": [f"Learner_{i}" for i in range(n_users)],
    "Age": np.random.normal(loc=29, scale=8.5, size=n_users).clip(14, 65).astype(int),
    "Gender": np.random.choice(["Female", "Male", "Non-Binary", "Prefer not to say"], size=n_users, p=[0.48, 0.46, 0.04, 0.02])
})

# Courses Table
courses = pd.DataFrame({
    "CourseID": [f"CRS_{100 + i}" for i in range(n_courses)],
    "CourseName": [f"Course Title {i}" for i in range(n_courses)],
    "CourseCategory": np.random.choice(["Data Science", "Software Engineering", "Business & Finance", "Design & UX", "Healthcare"], size=n_courses),
    "CourseType": np.random.choice(["Self-Paced", "Instructor-Led", "Bootcamp Short"], size=n_courses),
    "CourseLevel": np.random.choice(["Beginner", "Intermediate", "Advanced"], size=n_courses)
})

# Transactions Table
transactions = pd.DataFrame({
    "TransactionID": [f"TXN_{10000 + i}" for i in range(n_transactions)],
    "UserID": np.random.choice(users["UserID"], size=n_transactions),
    "CourseID": np.random.choice(courses["CourseID"], size=n_transactions),
})

# 2. Join Datasets Together (Data Integration step)
df_merged = transactions.merge(users, on="UserID", how="inner").merge(courses, on="CourseID", how="inner")

# 3. Segment Learners into Age Bands (Age Demographic Analysis)
bins = [0, 17, 25, 35, 45, 100]
labels = ["<18", "18–25", "26–35", "36–45", "45+"]
df_merged["AgeGroup"] = pd.cut(df_merged["Age"], bins=bins, labels=labels, right=True)

# Preview the first 5 rows of your integrated data
df_merged.head()
```

```python
# 1. Calculate Core KPIs for the Executive Summary / Research Paper
total_enrollments = len(df_merged)
unique_learners = df_merged["UserID"].nunique()
popular_category = df_merged["CourseCategory"].mode()[0]
avg_learner_age = int(df_merged["Age"].mean())

print(f"--- EDU-PRO ANALYTICS SUMMARY ---")
print(f"Total Enrollments: {total_enrollments}")
print(f"Unique Learners: {unique_learners}")
print(f"Most Popular Course Category: {popular_category}")
print(f"Average Learner Age: {avg_learner_age} years old\n")

# 2. Visual 1: Enrollments by Course Category
category_counts = df_merged["CourseCategory"].value_counts().reset_index()
category_counts.columns = ["CourseCategory", "Enrollments"]

fig_cat = px.bar(
    category_counts, 
    x="CourseCategory", 
    y="Enrollments", 
    color="CourseCategory",
    title="Course Popularity by Category",
    template="plotly_dark"
)
fig_cat.show()

# 3. Visual 2: Age Demographics Segmentation
age_counts = df_merged["AgeGroup"].value_counts().sort_index().reset_index()
age_counts.columns = ["AgeGroup", "Count"]

fig_age = px.pie(
    age_counts, 
    names="AgeGroup", 
    values="Count", 
    title="Learner Distribution across Age Demographics",
    hole=0.4,
    template="plotly_dark"
)
fig_age.show()
```
